```python
df = pd.read_csv("Student Marks.csv")
display(df)
```

```python
# Count missing values 
print(df.isnull().sum())
```

```python
# Checking duplicate rows
print(df.duplicated())
print("Total duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print(df.duplicated().sum())
```

```python
# Replace marks greater than 100 or less than 0 with subject average

subjects = ["Computer", "English", "Maths", "Programming", "Discrete"]
for subject in subjects:
    average = df[(df[subject] >= 0) & (df[subject] <= 100)][subject].mean()
    invalid = df[(df[subject] > 100) | (df[subject] < 0)]
    
    if len(invalid) > 0:
        print("Fixed invalid marks in", subject, "replaced with average", round(average, 2))
    
    df.loc[df[subject] > 100, subject] = average
    df.loc[df[subject] < 0, subject] = average

display(df)
```

```python
#Filling missing value with subjects average
for subject in subjects:
    df[subject] = df[subject].fillna(round(df[subject].mean()))
display(df)
```

```python
#Convert all subject columns to integers
for subject in subjects:
    df[subject] = df[subject].astype(int)

display(df)
```

```python
#Calculate Average Marks of Each Student
df["Average"] = ( df["Computer"] + df["English"] + df["Maths"] + df["Programming"] + df["Discrete"] ) / 5
display(df)
```

```python
# Find the student with the highest average
highest = df[df["Average"] == df["Average"].max()]
print("Student with the highest average are:")
print(highest)
# Find the student with the lowest average
lowest = df[df["Average"] == df["Average"].min()]

print("Student With the lowest average is:")
print(lowest)
```

```python
# Mean of every subject
print("Mean")
for subject in subjects:
    print(subject, "=", np.mean(df[subject]))
    
# Median 
print("Median")
for subject in subjects:
    print(subject, "=", np.median(df[subject]))

# Standard deviation 

print("Standard Deviation")
for subject in subjects:
    print(subject, "=", np.std(df[subject]))
```

```python
# Add Pass/Fail column
df["Result"] = np.where(df["Average"] >= 40, "Pass", "Fail")
display(df)
```

```python
print("Total Students =", len(df))

passed = len(df[df["Result"] == "Pass"])
print("Passed Students =", passed)

failed = len(df[df["Result"] == "Fail"])
print("Failed Students =", failed)

print("Subject wise Class Average")
for subject in subjects:
    print(subject, "=", round(df[subject].mean(), 2))
```

```python
df = df.sort_values(by="Average", ascending=False)
print("Students Sorted by Average Marks")
display(df)
```
